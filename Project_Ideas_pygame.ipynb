# Project Beer Game 2.0 — Single-Machine Playable Simulator

## What Is the Beer Game?

The **Beer Distribution Game** (or simply the *Beer Game*) is a classic supply chain management simulation originally developed at MIT in the 1960s. Its purpose is to demonstrate how **local decision-making, information delays, and order variability** can lead to large swings in inventory and backlog levels — a phenomenon known as the **bullwhip effect**.

In the traditional classroom version, four players represent the four stages of a supply chain:

1. **Retailer** — receives customer demand and orders from the Wholesaler.
2. **Wholesaler** — orders from the Distributor.
3. **Distributor** — orders from the Factory.
4. **Factory** — produces and ships beer to the Distributor.

Each week, every player must decide **how much beer to order** from their upstream partner, based only on their local information: incoming orders, incoming shipments, on-hand inventory, and backlog. Shipments and orders are delayed by several weeks, simulating real supply chain lead times. Even when customer demand is stable, small changes can create massive oscillations in the system.

The key learning objective is to **see and measure** how small, rational local decisions can produce instability in the global system — and to experiment with ways to stabilize it (better policies, information sharing, forecasting, etc.).

---

## Project Overview

You will build a **single-machine, playable Beer Game** that models the four supply chain tiers using **pygame** for visualization. The game supports **human and AI control** per tier (Retailer, Wholesaler, Distributor, Factory). Players can take turns or control a single tier while the others follow automated policies.

You will run several demand scenarios, log all decisions, analyze the bullwhip effect, and reflect on how control parameters affect performance.

As a **voluntary** add-on, you can extend it into a multi-user networked version.

---

## Game Rules

**Supply chain:** Four tiers in series: Retailer (R) → Wholesaler (W) → Distributor (D) → Factory (F).

**Weekly sequence of events (all tiers):**

1. **Receive shipments** due to arrive this week.
2. **Satisfy downstream demand** (for Retailer, this is customer demand). Unmet demand becomes backlog.
3. **Update inventory and backlog** after shipments.
4. **Place an order** to upstream partner (decision variable (Q_t \ge 0)).
5. **Advance pipelines:** orders and shipments each have a 1-week delay; Factory has an additional 1-week production delay.

**Delays:**

* Order information delay: 1 week
* Shipment delay: 1 week
* Factory production delay: 1 week

**Initial conditions:**

* On-hand inventory: 12 units; backlog: 0
* Shipment pipeline prefilled: 4 units arriving at week 1 and week 2 for each link

**Weekly costs:**

* Holding cost: 1 € per on-hand unit
* Backlog cost: 5 € per backordered unit

Each tier sees **only** its own data (inventory, backlog, incoming shipment, incoming order). There is no global dashboard.

---

## Play Modes

* **Single-player:** One human controls one tier (typically the Retailer); other tiers use AI policies.
* **Local multiplayer:** Up to four humans share one computer and take turns entering orders.
* **Mixed:** Any combination of human and AI control per tier.

Toggles in the pygame UI determine which policy controls each tier.

---

## Scenarios

Simulate **25 weeks** for each scenario. Record random seeds and parameters in your README.

1. **Step Change:** Demand = 4 (weeks 1–4), then 8 (weeks 5–25).
2. **Random (Poisson):** Demand ~ Poisson(λ = 6) each week.
3. **Trend + Noise:** Demand_t = round(4 + 0.4·t + ε_t), ε_t ~ N(0, 1.5), clipped ≥ 0.

---

## Implementation Tasks

1. **Core Simulation (pure Python):**

   * Represent state per tier (on_hand, backlog, order_pipeline, shipment_pipeline).
   * Implement the weekly update cycle with deterministic results given a seed.
   * Calculate holding, backlog, and cumulative costs.
2. **Policies:**

   * **Base-Stock Policy:** order up to target level S.
   * **PI Controller:** (Q_t = \alpha(\text{target} − \text{net stock}) + \beta \cdot \text{backlog} + \gamma).
   * Allow parameter tuning through UI or config.
3. **pygame UI:**

   * Visual panels per tier showing key state variables.
   * Control buttons: advance week, speed toggle, human/AI mode switch.
   * Log panel showing weekly events.
   * Save/load state to JSON.
4. **Logging:** Write one CSV row per tier per week with:

   ```csv
   run_id, scenario, seed, week, tier, demand_observed, order_received, order_placed,
   shipment_received, shipment_sent, on_hand_end, backlog_end,
   pipe_orders_next, pipe_shipments_next, holding_cost, backlog_cost, total_cost_cum
   ```

---

## Analysis Notebook (`BeerGame_Analysis.ipynb`)

For each scenario and configuration (human vs. AI):

1. **Summary table:** total cost, average inventory/backlog, service level, bullwhip ratio per tier (Var[orders]/Var[demand]).
2. **Plots:** inventory, backlog, orders, and shipments per tier over time.
3. **Bullwhip visualization:** bar chart comparing amplification by tier.
4. **Policy comparison:** Base-Stock vs. PI vs. Human control.
5. **Reflection (≤ 1 page):** What caused the bullwhip? How did control policies change system stability? How did your actions differ from algorithmic policies?

# Project Production Line Simulator — Single‑Machine Playable

## What Is the Production Line Simulator?

The **Production Line Simulator** models a factory line with multiple machines/stations, buffers, and stochastic processing. Jobs flow through a predefined routing (e.g., M1 → M2 → M3). You will explore how **variability, bottlenecks, setups, and breakdowns** create queues and impact **throughput, WIP, and cycle time** (Little’s Law: WIP = Throughput × Cycle Time). The simulator makes these dynamics **visible** and **measurable** and lets you test dispatching and scheduling rules.

---

## Project Overview

Build a **single‑machine, playable simulator** in **pygame** backed by a pure‑Python simulation core. The UI shows machines, buffers, and jobs as they move. You will implement baseline **dispatching/scheduling policies**, run scenarios, log all events, and analyze KPIs in a Jupyter Notebook. This Phase 1 version focuses on local play (one screen). A networked/multi‑user version can follow in Phase 2.

---

## Learning Goals

* Implement a discrete‑time or event‑driven production line with queues, setups, and failures.
* Diagnose bottlenecks and quantify **throughput, WIP, cycle time**, and **utilization**.
* Compare dispatching/scheduling rules (FIFO, SPT, EDD, CONWIP/Kanban, etc.).
* Design one **improvement** (e.g., buffer sizing, batching, setup reduction, reactive rescheduling) and evaluate it rigorously.

---

## System Definition (Phase 1 Rules)

**Line:** 3–5 machines in series (e.g., M1, M2, M3, optional M4/M5). Each machine has a finite input buffer (queue) with capacity Bᵢ (can be ∞ for baseline). Jobs follow a fixed routing.

**Jobs:** Each job j has a processing time pⱼ, due date dⱼ (for tardiness KPIs), and (optionally) a family/type that determines **sequence‑dependent setup times** s(type_prev → type_next) on each machine.

**Time:** Choose **discrete‑event** (recommended) or **fixed‑timestep discrete‑time**. Either way, the following events exist:

* **Arrival** of a new job to the system (Poisson/renewal or a fixed order list).
* **Start Processing** on a machine when idle and a job is available.
* **Finish Processing** after pⱼ (plus setup if required) → transfer to downstream buffer; if buffer full, the upstream machine becomes **blocked**.
* **Breakdown/Repair** on machines (MTBF/MTTR or geometric/exponential) → machine **starved/blocked** interactions matter.

**Breakdowns (optional in baseline, required in one scenario):** When a machine fails, any processing is paused; resume after repair time.

**Setups (optional in baseline, required in one scenario):** Changing job family incurs setup time before processing.

**Work release:** Either **open queue** (arrivals exogenous) or **WIP‑controlled** (e.g., **CONWIP**: release a new job when one completes at the last machine).

**KPIs (tracked continuously and per run):**

* **Throughput** (jobs/time), **Average WIP**, **Average Cycle Time** (arrival→completion), **Service Level** (on‑time delivery %), **Average Tardiness**, **Utilization** per machine, **Starvation/Blocking time** shares, **OEE** (optional), and **Setup time fraction**.

**Costs (for a unified objective):**

* WIP holding cost: 0.2 € per job per minute (or per tick).
* Tardiness penalty: 1 € per minute late (0 if on time).
* Setup cost: 0.5 € per minute of setup.
  (You may rescale if you use discrete ticks; keep costs linear for Phase 1.)

---

## Play Modes

* **Single‑player:** Player can pause and change policies, buffer limits, or release rules mid‑run.
* **Local team (one screen):** One student drives; teammates propose policy changes.
* **Mixed human/AI:** Policy can auto‑select jobs; human can override on a given machine.

The UI must provide toggles/selectors for policy per machine and a global speed control (pause/1×/4×).

---

# Project Delivery Route Challenge — Single‑Machine Playable 

## What Is the Delivery Route Challenge?

The **Delivery Route Challenge** is a last‑mile logistics simulator. You manage a fleet of vehicles serving customer stops on a city network under constraints such as **time windows**, **vehicle capacity**, and **maximum route duration**. Travel times depend on the road graph (and may vary by time of day). The goal is to plan and adapt routes to maximize **on‑time deliveries** while minimizing **distance, lateness, and operating cost**.

This project makes core Vehicle Routing Problem (VRP) ideas **visible** and **interactive** and lets you compare classical heuristics with human decisions.

---

## Project Overview

Build a **single‑machine, playable simulator** in **pygame** with a pure‑Python routing core. The UI shows a city map, depots, vehicles, and stops; you can **drag routes**, **toggle policies**, and **inject incidents** (road closures). You will run scenarios, log results, and analyze KPIs in a Jupyter Notebook. Phase 1 focuses on local play; a networked/multi‑user or map‑backed version can follow in Phase 2.

---

## Learning Goals

* Model and solve simplified VRP variants (capacity, time windows, route duration).
* Implement constructive and local‑search heuristics and compare against human edits.
* Analyze operational trade‑offs: on‑time performance vs. distance and fleet size.
* Design and evaluate one **improvement** (e.g., dynamic re‑routing, insertion for new orders, or simple predictive ETAs).

---

## System Definition 

**Network:** Undirected road graph (G = (V, E)) with non‑negative edge weights (distance or travel time). For Phase 1 you may use a generated grid graph (e.g., 20×20) with Manhattan distances; optional time‑of‑day multiplier.

**Entities:**

* **Depot(s):** starting/ending node(s) for vehicles.
* **Customers/Stops:** nodes with demand `d_i` (units), service time `s_i`, and optional **time window** `[a_i, b_i]`.
* **Vehicles:** capacity `Q`, max route duration `T_max`, and average speed `v` (if using time‑based travel).

**Schedule semantics:** A route is a sequence Depot → stops → Depot. Arrival after `b_i` incurs **lateness** `L_i = max(0, t_arrive − b_i)`; arriving before `a_i` waits until `a_i`.

**Costs & KPIs:**

* **Distance (or travel time)** total; **lateness penalty** per minute; **route fixed cost** per vehicle used.
* KPIs: % deliveries on time, total distance, average lateness of late deliveries, # vehicles used, route duration stats.

**Dynamics (optional but recommended):** During runtime you may **add a new order** or **close an edge** to test re‑routing.

---

## Play Modes 

* **Single‑player:** Human plans/edits routes; algorithms can propose initial solutions.
* **Local team:** One student drives; teammates advise and toggle heuristics.
* **Mixed:** Any subset of vehicles is AI‑controlled (auto‑improve), others human‑edited.

UI must provide: pause/1×/4× speed, “build initial routes”, “improve routes”, and interactive drag‑and‑drop of stops between routes.

---

## Scenarios (Run each with fixed seeds; 60–120 stops)

Run at least **three** scenarios:

1. **Capacitated VRP (CVRP, S₀):** Single depot, no time windows; `Q` limits; evaluate distance and vehicles used.
2. **VRPTW (S₁):** Single depot; hard time windows with moderate slack; service times non‑zero; evaluate on‑time %.
3. **Dynamic Incident (S₂):** Start from S₁; at simulated time τ close 5–10% of edges (or increase travel times), or inject 5 new orders; evaluate re‑routing impact.

Document all parameters: `Q`, `T_max`, `v`, time‑window tightness, service times, seeds.

---

## Implementation Tasks

1. **Routing Core (pure Python):**

   * Shortest paths on the graph (precompute all‑pairs on small graphs or use Dijkstra on demand).
   * Feasibility checks: capacity, time windows, route duration.
   * Cost function: distance + lateness penalty + route fixed cost; compute KPIs.
2. **Heuristics (≥ 2 constructive + 1 local search):**

   * **Constructive:** Clarke–Wright savings; Sweep (polar angle clustering); Nearest‑neighbor insertion.
   * **Local search:** 2‑opt (intra‑route), Or‑opt (segment relocation), cross‑exchange (inter‑route), or relocate‑insert with best‑insertion.
   * **Improvement (choose ≥ 1):** time‑window‑aware insertion; destroy‑and‑repair (random removal + greedy insert); rolling‑horizon re‑insertion for dynamic orders.
3. **pygame UI:**

   * Map rendering (grid or node/edge plot), depot(s), stops colored by route; vehicles shown moving along routes (optional animation).
   * Panels: route list with capacity/time utilization bars; per‑stop ETA and lateness indicator; global KPIs.
   * Interactions: drag stop to a different route/position (snap to best insert), trigger heuristics, toggle edges open/closed, add a new order.
4. **Logging:**

   * **Routes CSV (`routes.csv`)** at key milestones (initial, after improvements, final):

     ```csv
     run_id,stage,route_id,seq_index,node_id,arrive,depart,load_after,feasible,late_minutes
     ```
   * **KPIs CSV (`kpi.csv`)** after each improvement step and at simulation end:

     ```csv
     run_id,stage,total_distance,on_time_pct,avg_late_minutes,vehicles_used,avg_route_duration,max_route_duration,cost
     ```
   * **Incidents CSV (`events.csv`)** for dynamic changes:

     ```csv
     run_id,time,event_type,edge_or_node,detail
     ```

---

## Analysis Notebook (`DeliveryRoute_Analysis.ipynb`)

For each scenario:

1. **Run Summary Table:** total distance, on‑time %, average lateness, vehicles used, average/max route duration, total cost.
2. **Route Visuals:** final route plot with color per route and stop labels; optional before/after overlay.
3. **Heuristic Ablation:** constructive vs. +2‑opt vs. +Or‑opt, etc.; show improvement curve (cost vs. iteration).

# Project Warehouse Picker Game — Single‑Machine Playable

## What Is the Warehouse Picker Game?

The **Warehouse Picker Game** simulates order picking operations in a warehouse environment. Players manage one or more **pickers** moving through aisles to collect products for customer orders. The challenge is to **minimize travel distance and time**, **avoid congestion**, and **balance workload** between pickers. The simulator illustrates how **layout design**, **routing policies**, and **order batching** affect efficiency and throughput.

This project connects directly to warehouse logistics and operations management concepts — especially **order picking optimization**, **routing heuristics**, and **queueing under spatial constraints**.

---

## Project Overview

Develop a **single‑machine, playable simulator** using **pygame** with a modular Python back‑end. The UI should visualize the warehouse grid (racks, aisles, pick locations, docks, pickers). Players or algorithms can plan picker paths, release orders, and manage congestion. You will simulate multiple demand patterns, compare routing policies, log key metrics, and analyze results in a Jupyter Notebook.

First emphasize single‑player operation; later phases may extend to multi‑user or multi‑warehouse coordination.

---

## Learning Goals

* Understand how layout, batching, and routing decisions impact throughput and picker workload.
* Implement classic **picker routing heuristics** (S‑shape, Largest Gap, Return, Composite) and evaluate their performance.
* Analyze the effect of **order batching** and **congestion** on total travel time.
* Build a foundation for further optimization, e.g., dynamic zoning or cooperative multi‑picker scheduling.

---

## System Definition

**Warehouse Layout:** Rectangular grid with defined aisles and storage locations. Each cell can be:

* **Rack** (storage, not walkable)
* **Aisle** (walkable)
* **Dock/Start** (picker’s base station or order consolidation area)

**Orders:** Each order has a list of pick locations (cells) and total weight/volume. Orders arrive according to predefined scenarios.
**Pickers:** Move one cell per tick; each has a carrying capacity limit (weight or number of items). Pick time per item = 1 tick. When full or all assigned orders complete, return to dock.

**Routing:** Pickers must find a path visiting all assigned pick locations and returning to the dock. Travel only through aisles.
**Routing Policies:**

* **S‑shape:** traverse aisles fully for items within that aisle.
* **Largest Gap:** skip large gaps, turn back sooner.
* **Return:** enter aisle to deepest required pick, then return.
* **Composite:** hybrid minimizing total distance.

**KPIs:** Total travel distance/time, total orders completed, average picker utilization, average lateness (if time windows exist), congestion index (mean waiting time at intersections).

---
